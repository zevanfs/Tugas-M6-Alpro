```python
#Nomor 1
usia = int(input("Masukkan usia Anda:"))
departemen = input("Masukkan asal Departemen Anda:")

departemen = departemen.title()
syarat_asisten = (usia >= 19) and ("Statistika" in departemen)

print(f"Usia       : {usia}")
print(f"Departemen : {departemen}")
print(f"Apakah formulir pendaftaran diterima? {syarat_asisten}")
```

```python
#Nomor 2
data_pelanggan = ("Zeva", "Premium")
berat_kardus = [9, 10, 20, 26]

(nama, status_member) = data_pelanggan
total_berat = sum(berat_kardus)
syarat_VIP = (status_member == "Premium") or (total_berat > 100)

print(f"Nama pelanggan     : {nama}")
print(f"Status member      : {status_member}")
print(f"Total berat kardus : {total_berat} kg")
print(f"Apakah pengiriman masuk jalur prioritas VIP? {syarat_VIP}")
```

```python
#Nomor 3
ipk = float(input("Masukkan IPK Anda:"))
riwayat_nilai = ["A", "A", "AB", "A", "B", "AB"]

lolos_beasiswa = (ipk >= 3.50) and ("E" not in riwayat_nilai)

print(f"IPK           : {ipk}")
print(f"Riwayat nilai : {riwayat_nilai}")
print(f"Apakah lolos beasiswa? {lolos_beasiswa}")
```
